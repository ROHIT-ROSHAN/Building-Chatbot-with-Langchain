```python
import os
from dotenv import load_dotenv
load_dotenv()
## Loading all the env variables

groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
from langchain_openai import ChatOpenAI
from langchain_groq import ChatGroq
model=ChatGroq(model="openai/gpt-oss-120b",groq_api_key=groq_api_key )
```

```python
from langchain_core.messages import HumanMessage
model.invoke(
    [
        HumanMessage(content="Hi My name is Rohit, I am a chief AI Engineer"),

    ]
)
```

```python
from langchain_core.messages import AIMessage
model.invoke(
    [
        HumanMessage(content="Hi My name is Rohit, I am a chief AI Engineer"),
        AIMessage(content="Hello Rohit! 👋\n\nGreat to meet a chief AI Engineer. How can I help you today? Whether you’re looking for technical insights, brainstorming ideas, troubleshooting a model, or anything else—just let me know!"),
        HumanMessage(content="Hey what is my name and what do I do")
    ]
)
```

```python
### Message History
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory


store  = {}
def get_session_history(session_id:str)->BaseChatMessageHistory:
    if session_id not in store:
        store[session_id]= ChatMessageHistory()
    return store[session_id]

with_message_history = RunnableWithMessageHistory(model , get_session_history)
```

```python
config = {"configurable":{"session_id":"chat1"}}
```

```python
response=with_message_history.invoke(
    [
        HumanMessage(content="Hi My name is Rohit, I am a chief AI Engineer"),
    ],
    config=config
)
```

```python
response.content
```

```python
response=with_message_history.invoke(
    [
        HumanMessage(content="Hi What is my name, and what do I do?"),
    ],
    config=config
)
```

```python
response.content
```

## Prompt Template

```python
from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder
prompt = ChatPromptTemplate.from_messages(
    [
        ("system","You are a helpful assistant. Answer all the questions in the best of your ability"),
        MessagesPlaceholder(variable_name="messages"),
        
    ]
)
chain = prompt|model
```

```python
chain.invoke({"messages":[HumanMessage(content="Hi, My name is Rohit")]})
```

```python
with_message_history = RunnableWithMessageHistory(chain,get_session_history)
```

```python
config = {"configurable":{"session_id":"chat3"}}
response = with_message_history.invoke(
    [
        HumanMessage(content=" ")     
    ],
    config=config,
)
```

```python
response.content
```

## Managing the Conversation History

```python
from langchain_core.messages import SystemMessage, trim_messages
# trim_messages - helps how many tokens can we send to the model
```

```python
trimmer = trim_messages(
    max_tokens=40,
    strategy="last",
    token_counter=model, 
    include_system=True, # Include the system message
    allow_partial=False, # Allow partial message
    start_on="human" # where it needs to start conversation
)
messages = [
    SystemMessage(content="you are a good assistant"),
    HumanMessage(content="Hi I am Bob"),
    AIMessage(content="HI"),
    HumanMessage(content="I like vanilla ice cream"),
    AIMessage(content="Nice"),
]
trimmer.invoke(messages)
```

```python
from operator import itemgetter

from langchain_core.runnables import RunnablePassthrough

chain=(
    RunnablePassthrough.assign(messages=itemgetter("messages")|trimmer)|prompt|model
)
```

```python
response=chain.invoke(
    {
    "messages":messages+ [HumanMessage(content="What icecream do I like?")],
    "language":"English"
    }
)
response.content
```

```python
# Lets wrap this in message history
with_message_history = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="messages",
)
config={"configurable":{"session_id":"chat5"}}
```

```python
response=with_message_history.invoke(
    {
    "messages":messages+ [HumanMessage(content="What icecream do I like?")],
    "language":"English"
    },
    config=config
)
response.content
```
